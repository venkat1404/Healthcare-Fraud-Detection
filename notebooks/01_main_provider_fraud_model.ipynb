## Responsible AI Notes

- False positives can delay legitimate claims.
- False negatives can let fraud payments go through.
- Claims above $10,000 flagged as suspicious should require human review before denial.
- Predictions below 80% confidence should require human verification.
- Providers with no prior fraud history should not be denied automatically.
- The model should assist investigators, not replace them.
